```python
import pandas as pd
import os
files=os.listdir()
csv_files=[f for f in files if f.startswith("DL")and f.endswith(".csv")]
print(csv_files)
df_list=[]
for file in csv_files:
    print("loading:",file)
    temp_df=pd.read_csv(file)
    df_list.append(temp_df)
final_df=pd.concat(df_list,ignore_index=True)
print(final_df.shape)
```

```python
final_df['timestamp']=pd.to_datetime(final_df['timestamp'])
```

```python
final_df.info()
```

```python
final_df.drop(columns=["vws" , "eth_benzene" , "mp_xylene" , "o_xylene" , "xylene"],inplace=True) 
```

```python
final_df.info()
```

```python
print(final_df.isnull().sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(final_df["pm10"].dropna())
plt.title("pm10 Boxplot")
plt.show()
```

```python
final_df["pm2.5"]=final_df["pm2.5"].fillna(final_df["pm2.5"].median())
final_df["pm10"]=final_df["pm10"].fillna(final_df["pm10"].median())
```

```python
num_cols=final_df.select_dtypes(include=["float64","int64"]).columns
final_df[num_cols]=final_df[num_cols].fillna(final_df[num_cols].median())
```

```python
print("Before", final_df.shape)
```

```python
final_df=final_df.drop_duplicates()
print("After", final_df.shape)
```

```python
final_df["timestamp"]=pd.to_datetime(final_df["timestamp"])
final_df["year"]=final_df["timestamp"].dt.year
final_df["month"]=final_df["timestamp"].dt.month
final_df["day"]=final_df["timestamp"].dt.day
final_df["hour"]=final_df["timestamp"].dt.hour
```

```python
final_df[["timestamp","year","month","day","hour"]].head()
```

```python
sample_df=final_df.sample(n=200000,random_state=42)
```

```python
sample_df.shape
```

```python
x=sample_df.drop(["pm2.5","timestamp"],axis=1)
y=sample_df["pm2.5"]
```

```python
x.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
print("R2 score: ",r2_score(y_test, y_pred))
print("MAE: ",mean_absolute_error(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf=RandomForestRegressor(n_estimators=50, random_state=42)
```

```python
rf.fit(x_train, y_train)
```

```python
rf_pred=rf.predict(x_test)
```

```python
print("R2 Score: ", r2_score(y_test, rf_pred))
print("MAE: ", mean_absolute_error(y_test, rf_pred))
```

```python
importance=rf.feature_importances_
```

```python
feature_importance =pd.DataFrame({"Feature" : x.columns, "Importance" : importance})
```

```python
feature_importance=feature_importance.sort_values(by="Importance", ascending=False)
print(feature_importance)
```

```python
feature_importance.plot(x="Feature", y="Importance", kind="barh", figsize=(10,6), color="red",edgecolor="green")
plt.title("Feature Importance")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(y_test, rf_pred, alpha=0.5, color="red")

plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         color="blue", linestyle="--", linewidth=2)

plt.xlabel("Actual PM2.5")
plt.ylabel("Predicted PM2.5")
plt.title("Actual vs Predicted (Random Forest)")
plt.grid(True)
plt.show()
```
